```python
import pandas as pd
import glob
import json
import os

LANGUAGES = ["PB NATIVO", "ESPANHOL", "FRANCÊS", "GALEGO", "INGLÊS", "MANDARIM", "RUSSO", "ÁRABE"]
N_LANGUAGES = len(LANGUAGES) 

TASKS = ["CAPACITAR", "COMPARTILHAR", "EXPLICAR", "EXPLORAR", "FAZER", "RECOMENDAR", "RECRIAR", "RELATAR"]
N_TASKS = len(TASKS)
```

```python
files = glob.glob("./grammar_swap/causalLM/*.json", recursive=True)
files.sort()
files
```

```python
data = {
    "Model": [],
    "Task": [],
    "Language": [],
    "Tokens/Word": [],
    "Perplexity": [],
    "Pseudo-Perplexity": [],
    "BPB": [],
}

for file_path in files:
    with open(file_path, "r") as file:
        json_data = json.load(file)
    
    model_name = os.path.basename(file_path).replace(".json", "")

    for task in TASKS:
        for language in LANGUAGES:

            try:
                curr_data = json_data[task][language]

                data["Model"].append(model_name)
                data["Task"].append(task)
                data["Language"].append(language)
                
                token_per_word = curr_data.get("avg_tokens_per_word", None)
                ppl = curr_data.get("avg_ppl", None)
                pseudo_ppl = curr_data.get("avg_pseudo_ppl", None)
                bpb = curr_data.get("avg_bpb", None)

                data["Tokens/Word"].append(token_per_word)
                data["Perplexity"].append(ppl)
                data["Pseudo-Perplexity"].append(pseudo_ppl)
                data["BPB"].append(bpb)
            except:
                pass

print(data)
df = pd.DataFrame(data)
df.to_csv("grammar_swap_causalLM_results_compiled.csv", index=False)
```
