```python
from glob import glob
import os
import json
```

```python
files = glob("Albertina900m/*.json")
files.sort()
files
print(files)
data = dict()

for f in files:
    dataset = os.path.basename(f).split("-")[-1].replace(".json", "")
    with open(f, "r") as f:
        temp = json.load(f)
        data[dataset] = temp

with open("Albertina900m.json", "w") as file:
    json.dump(data, file, indent=4)
```
