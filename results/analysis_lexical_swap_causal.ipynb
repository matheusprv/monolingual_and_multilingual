```python
import pandas as pd
df = pd.read_csv("lexical_swap_causalLM_results_compiled.csv")

df = df.dropna(subset=["Perplexity"])
df = df.drop(columns=["Pseudo-Perplexity"])
```

```python
df[['Tokens/Word', 'Perplexity', 'BPB']].corr(method='spearman')
```

```python
metrics = ["Tokens/Word", "Perplexity", "BPB"]

pb_ref = df[df["Language"] == "PB NATIVO"][
    ["Model", "Task"] + metrics
].rename(columns={m: f"{m}_PB_NATIVO" for m in metrics})

df = df.merge(pb_ref, on=["Model", "Task"], how="left")

for metric in metrics:
    df[f"{metric} % vs PB NATIVO"] = (
        (df[metric] / df[f"{metric}_PB_NATIVO"] - 1) * 100
    )

df = df.drop(columns=[f"{m}_PB_NATIVO" for m in metrics])

df = df[df['Language'] != "PB NATIVO"]

df = df.round(2)
```
