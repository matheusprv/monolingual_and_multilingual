```python
def extract_corpora_data(corpora):
    word_to_lemma = dict()
    lemma_to_words = dict()

    for corpus in corpora:
        # read and parse corpus
        with open(corpus) as file:
            data = file.read()

        sentences = parse(data)
    
        # extract the words and lemmas
        for sentence in sentences:
            for token in sentence:
                form = token['form']
                lemma = token['lemma']

                if form not in word_to_lemma:
                    word_to_lemma[form] = lemma

                if lemma not in lemma_to_words:
                    lemma_to_words[lemma] = set()
                
                lemma_to_words[lemma].add(form.lower())


    lemma_to_words = {k: list(v) for k, v in lemma_to_words.items()}

    return word_to_lemma, lemma_to_words


corpora = [
    "../../../corpora/Universal Dependencies/UD_Portuguese-Bosque/pt_bosque-ud-dev.conllu",
    "../../../corpora/Universal Dependencies/UD_Portuguese-Bosque/pt_bosque-ud-test.conllu",
    "../../../corpora/Universal Dependencies/UD_Portuguese-Bosque/pt_bosque-ud-train.conllu",
    
    "../../../corpora/Universal Dependencies/UD_Portuguese-GSD/pt_gsd-ud-dev.conllu",
    "../../../corpora/Universal Dependencies/UD_Portuguese-GSD/pt_gsd-ud-test.conllu",
    "../../../corpora/Universal Dependencies/UD_Portuguese-GSD/pt_gsd-ud-train.conllu",

    "../../../corpora/Universal Dependencies/UD_Portuguese-Porttinari/pt_porttinari-ud-dev.conllu",
    "../../../corpora/Universal Dependencies/UD_Portuguese-Porttinari/pt_porttinari-ud-test.conllu",
    "../../../corpora/Universal Dependencies/UD_Portuguese-Porttinari/pt_porttinari-ud-train.conllu"
]

word_to_lemma, lemma_to_words = extract_corpora_data(corpora)
```

```python
def corrupt_sentence(sentence):
    tokens = nltk.word_tokenize(sentence)

    corrputed_sentence = []

    for token in tokens:
        lemma = word_to_lemma.get(token, token)

        if lemma == token:
            corrputed_sentence.append(token)
        
        else:
            new_word = random.choice(lemma_to_words[lemma])
            corrputed_sentence.append(new_word)

    return " ".join(corrputed_sentence)
    
```
