```python
import json
import random
import nltk
```

```python
with open('word_to_lemma.json', 'r') as file:
    word_to_lemma = json.load(file)

with open('lemma_to_words.json', 'r') as file:
    lemma_to_words = json.load(file)
```

```python
def corrupt_sentence(sentence):
    tokens = nltk.word_tokenize(sentence)

    corrputed_sentence = []

    for token in tokens:
        lemma = word_to_lemma.get(token, token)

        if lemma == token:
            corrputed_sentence.append(token)
        
        else:
            new_word = random.choice(lemma_to_words[lemma])
            corrputed_sentence.append(new_word)

    return " ".join(corrputed_sentence)
```
