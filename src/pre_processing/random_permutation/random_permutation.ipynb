```python
import os 
import glob
import random

import spacy
nlp = spacy.load("pt_core_news_lg")
```

```python
SPACE_AFTER = {".", "..", "...", ",", ";", ":", "!", "?", "…", ")", "]", "}", "»", "”", "’", "%"}
SPACE_BEFORE = {"(", "[", "{", "«", "“", "‘"}


def permut_text(text: str):
    doc = nlp(text)

    chunk = []
    shuffled_text = ""

    total_tokens = len(doc)

    token = None
    for i in range(total_tokens):
        token = doc.__getitem__(i)

        if token.text in ('\n', '\n\n'):
            shuffled_text += token.text

        elif token.pos_ != "PUNCT":
            chunk.append(token.text)

        else:
            # Adding the shuffled text
            random.shuffle(chunk)
            shuffled_text += " ".join(chunk)

            chunk = [] 

            # Defining if there is a blank space before or after any punctuation
            if token.text in SPACE_BEFORE:
                if len(shuffled_text) == 0 or shuffled_text[-1] == " ":
                    shuffled_text += token.text
                else: 
                    shuffled_text += " " + token.text

            elif token.text in SPACE_AFTER:
                if i < total_tokens-1 and doc.__getitem__(i+1).pos_ == "PUNCT":
                    shuffled_text += token.text
                else: 
                    shuffled_text += token.text + " "
            else:
                shuffled_text += token.text
    
    # In case the last sentence does not end with a PUNCT
    if token.pos_ != "PUNCT":
        random.shuffle(chunk)
        shuffled_text += " ".join(chunk)

    return shuffled_text        
    
```

```python
texto = "Oi tati.. seja bem vinda"
permut_text(texto)
```

```python
def permut_corpus(file_name, target_folder="./"):

    new_corpus = ""
    new_corupus_file_name = f"{target_folder}/permuted_" + os.path.basename(file_name)
    
    with open(file_name, "r") as f:
        for line in f:
            line = line.strip()
            if len(line) > 0 and line[0] not in ("<", "---"):
                line = permut_text(line)
            new_corpus += line + "\n"

    with open(new_corupus_file_name, "w") as f:
        f.write(new_corpus)   

```

```python
SOURCE_FOLDER = "../../../corpora/GeneratedCorpus/PB NATIVO/"
TARGET_FOLDER = "../../../corpora/GeneratedCorpus/PB NATIVO - PERMUTED/"

os.makedirs(TARGET_FOLDER, exist_ok=True)

source_files = glob.glob(SOURCE_FOLDER+"*")
total_files = len(source_files)

for i, file_path in enumerate(source_files):
    print(f"{i+1}/{total_files} - {file_path}")
    permut_corpus(file_path, TARGET_FOLDER)

```
