```python
import os 
from nltk import tokenize
import random

import spacy
nlp = spacy.load("pt_core_news_lg")
```

```python
texto = """
Caros colegas, funcionários e professores do ICHS, 

Ontem li um recado disposto em uma das paredes do ICHS informando sobre a retirada dos cães que vivem no Campus. Não sei qual o envolvimento que cada um tem, seja com os animais que aqui habitam, seja conhecimento sobre o que o setor de Zoonose de Mariana faz com os cães que recolhem (mesmo havendo diversas testemunhas que alegam que TODOS os animais recolhidos anteriormente VOLTARAM DOENTES E FRACOS, alguns até morreram um pouco depois de voltarem para casa...)
 
O fato é que os cães que vivem no ICHS são cuidados por alunos, professores e funcionários do Instituto, e sua retirada, além de terrível para nós que zelamos pela vida e bem estar desses animais, e para os próprios cães que fazem do local sua CASA, será uma solução apenas momentânea. Isso porque, esses cães são (e continuarão sendo) abandonados no Campus, reproduzem-se indiscriminadamente, e não há sequer um projeto seja da Secretaria de Saúde do Município, seja da própria Prefeitura, ou da Universidade, que estimule, dentro e fora do Instituto, atitudes mais racionais, humanas ( e até baratas ), como a esterilização destes animais ( a castração, no caso ), a vacinação gratuita, o incentivo à posse responsável e a adoção destes. 

Sugiro portanto que NÃO PERMITAMOS a retirada dos cães, e que PRESSIONEMOS os órgãos competentes a medidas mais sadias, para nós e nossos animais: uma mudança de conduta, em lugar de um impulso estéril e inconsequente; e que pressionemos JÁ, pois que a data para a retirada dos cães é o próximo dia 15, data em que muitos alunos já estão de férias!!!

TODOS OS CÃES, desde que foram domesticados a milhares de anos, dependem inevitavelmente do laço com os seres humanos para sobreviver, o que nos torna RESPONSÁVEIS diretos sobre quaisquer danos às suas vidas.
"""
texto = texto.strip()
```

```python
SPACE_AFTER = {".", ",", ";", ":", "!", "?", "…", ")", "]", "}", "»", "”", "’", "%"}
SPACE_BEFORE = {"(", "[", "{", "«", "“", "‘"}


def mess_text(text: str):
    doc = nlp(text)

    chunk = []
    shuffled_text = ""

    total_tokens = len(doc)

    for i in range(total_tokens):
        token = doc.__getitem__(i)


        if token.text in ('\n', '\n\n'):
            shuffled_text += token.text

        elif token.pos_ != "PUNCT":
            chunk.append(token.text)

        else:
            # Adding the shuffled text
            random.shuffle(chunk)
            shuffled_text += " ".join(chunk)

            chunk = [] 

            # Defining if there is a blank space before or after any punctuation
            if token.text in SPACE_BEFORE:
                if shuffled_text[-1] == " ":
                    shuffled_text += token.text
                else: 
                    shuffled_text += " " + token.text

            elif token.text in SPACE_AFTER:
                if i < total_tokens-1 and doc.__getitem__(i+1).pos_ == "PUNCT":
                    shuffled_text += token.text
                else: 
                    shuffled_text += token.text + " "
            else:
                shuffled_text += token.text

    return shuffled_text        
    

print(mess_text(texto))
```
