```python
import pandas as pd
```

```python
df = pd.read_parquet("corpora.parquet")
df.head()
```

```python
df["scenario"].unique()
```

```python
GS_v2 = pd.read_parquet("GRAMMAR SWAP v2.parquet")
GS_v3_en = pd.read_parquet("GRAMMAR SWAP v3_en.parquet")
GS_v3_es = pd.read_parquet("GRAMMAR SWAP v3_es.parquet")

GS_v3_en = GS_v3_en.drop(columns=["text_id", "original_text", "stanza_tokens", "words_mapping"])
GS_v3_es = GS_v3_es.drop(columns=["text_id", "original_text", "stanza_tokens", "words_mapping"])
```

```python
new_corpora = pd.concat([df, GS_v2, GS_v3_en, GS_v3_es], ignore_index=True)
new_corpora = new_corpora.reset_index(drop=True)

new_corpora["text"] = new_corpora["text"].str.replace(r"\n+", " ", regex=True)
new_corpora
```

```python
new_corpora.to_parquet("corpora_v3.parquet")
```
